## 1. Algoritmo

1. Leitura dos dados  
   - Abre os arquivos que contém os dados de médias mensais e observações diárias que o SIDC (Solar influences Data Analysis Center) coletou e lê linha por linha.  
   - Converte os campos de data e contagem em números e armazena em dicionários organizados por ano e mês (e dia, quando é diário).  
   - Valores ausentes são tratados de forma explícita para não distorcer as estatísticas.

2. Cálculo dos extremos no período  
   - Para um intervalo definido pelo usuário, busca o máximo diário nos dados diários e o máximo e mínimo mensais nos dados mensais.  
   - Gera resultados mesmo se houver meses ou dias sem dados, avisando quando faltam observações.

3. Contagem de dias sem manchas  
   - Conta os dias com contagem igual a zero em cada mês a partir dos dados diários.

   - Aplica média móvel de 13 meses: pesos 0.5 para o mês inicial e final da janela e peso 1 para os 11 outros meses.  
   - Quando faltam meses, o denominador é ajustado para usar apenas observações válidas.


5. Detecção de ciclos  
   - Usa a série suavizada para localizar mínimos e máximos ou cruzamentos com a média do período.  
   - Oferece quatro métodos: mínimo‑a‑mínimo, máximo‑a‑máximo, cruzamento ascendente e cruzamento descendente.  
   - Adota uma tolerância numérica para evitar problemas com comparações de ponto flutuante e trata platôs (valores iguais em sequência).

6. Testes estatísticos  
   - Implementa teste t (Welch) para comparar médias entre dois ciclos.  
   - Implementa teste Qui‑Quadrado para comparar distribuição por categorias; antes de aplicar, verifica se as contagens esperadas são adequadas.  
   - Se as condições do Qui‑Quadrado não forem atendidas, aplica um teste alternativo (Kolmogorov-Smirnov) apropriado.

## 2.5 Metodologia

Explicação curta e direta dos principais métodos usados.

### Média Móvel Suavizada SIDC

A média suavizada do SIDC usa 13 meses: os meses nas extremidades da janela têm peso 0.5; os 11 meses centrais têm peso 1. No código, se algum mês estiver ausente, ele não entra no somatório e o denominador é ajustado para refletir só os valores disponíveis.

### Métodos de detecção de ciclos

- Mínimo a Mínimo: usa vales consecutivos na série suavizada para definir início e fim de ciclo (critério oficial do SIDC).
- Máximo a Máximo: usa picos consecutivos; é útil para estudar a duração da fase ativa.
- Cruzamento Ascendente: marca quando a curva suavizada cruza para cima a média do período.
- Cruzamento Descendente: marca quando a curva suavizada cruza para baixo a média do período.

O código deixa claro como tratar meses na fronteira entre ciclos para evitar ambiguidade.

### Testes Estatísticos Utilizados

Para comparar ciclos solares, foram aplicados três testes estatísticos:

- Teste t de Welch: compara as médias de duas amostras independentes, mesmo que tenham tamanhos ou variâncias diferentes. A estatística é calculada por:

$$
t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}
$$

Onde:
* $\bar{x}_1$ e $\bar{x}_2$: Médias das amostras (média de manchas solares dos dois ciclos comparados).
* $s_1^2$ e $s_2^2$: Variâncias amostrais dos dois grupos (medida de dispersão dos dados).
* $n_1$ e $n_2$: Tamanhos das amostras (número total de meses com dados em cada ciclo).

É usado para verificar se dois ciclos têm médias significativamente diferentes.

- Teste Qui‑Quadrado: compara a distribuição de frequências observadas com uma distribuição esperada. A fórmula é:

$$
\chi^2 = \sum_{i} \frac{(O_i - E_i)^2}{E_i}
$$

Onde:
* $O_i$: Frequência observada no intervalo $i$ (dados do ciclo específico sendo analisado).
* $E_i$: Frequência esperada no intervalo $i$ (baseada na distribuição histórica de todos os dados).

Foi usado para verificar se o perfil de atividade de um ciclo é compatível com o padrão histórico.

- Teste de Kolmogorov–Smirnov (KS): compara duas distribuições acumuladas e mede a maior diferença entre elas. É útil quando o Qui‑Quadrado não pode ser aplicado (por exemplo, quando há poucas observações por categoria).

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import math
from pathlib import Path
from typing import Dict, Tuple, List, Optional, Any
import logging

# Configuração básica de log
logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")

# Constantes Globais
LISTA_CICLOS_OFICIAIS = [
    (1, 1755, 2, 1766, 6), (2, 1766, 6, 1775, 6), (3, 1775, 6, 1784, 9),
    (4, 1784, 9, 1798, 5), (5, 1798, 5, 1810, 12), (6, 1810, 12, 1823, 5),
    (7, 1823, 5, 1833, 11), (8, 1833, 11, 1843, 7), (9, 1843, 7, 1856, 2),
    (10, 1856, 2, 1867, 3), (11, 1867, 3, 1878, 12), 
    (12, 1878, 12, 1890, 3),
    (13, 1890, 3, 1902, 1), (14, 1902, 1, 1913, 8), (15, 1913, 8, 1923, 8),
    (16, 1923, 8, 1933, 9), (17, 1933, 9, 1944, 2), (18, 1944, 2, 1954, 4),
    (19, 1954, 4, 1964, 10), (20, 1964, 10, 1976, 6), (21, 1976, 6, 1986, 9),
    (22, 1986, 9, 1996, 8), (23, 1996, 8, 2008, 12), (24, 2008, 12, 2019, 12),
    (25, 2019, 12, None, None)
]

EPS = 1e-9

def ano_mes_to_total_months(ano: int, mes: int) -> int:
    """
    Converte ano e mês para uma contagem absoluta de meses.
    Útil para cálculos de deslocamento (offset) temporal.
    """
    return ano * 12 + (mes - 1)

def total_months_to_ano_mes(total: int) -> Tuple[int, int]:
    """
    Converte uma contagem absoluta de meses de volta para (ano, mês).
    """
    return total // 12, (total % 12) + 1

def decimal_para_ano_mes(data_dec: float) -> Tuple[int, int]:
    """
    Converte data em formato decimal para (ano, mês).
    Evita erros de ponto flutuante usando math.floor.
    Exemplo: 1996.5 -> (1996, 7).
    """
    ano = int(math.floor(data_dec))
    frac = data_dec - ano
    mes = int(math.floor(frac * 12 + 1e-12)) + 1
    if mes > 12:
        ano += 1
        mes = 1
    return ano, mes

def ano_mes_para_decimal(ano: int, mes: int) -> float:
    """
    Converte (ano, mês) para representação decimal do ano.
    """
    return ano + (mes - 1) / 12.0

def ler_dados_mensais(nome_arquivo: str = "SN_m_tot_V2.0.txt") -> Dict[int, Dict[int, Tuple[float, Optional[float], Optional[float], Optional[int]]]]:
    """
    Lê o arquivo de dados mensais de manchas solares (formato SIDC).
    Retorna um dicionário aninhado: {ano: {mes: (fracao, num_manchas, desvio, nobs)}}.
    Retorna dicionário vazio em caso de erro de leitura.
    """
    dados = {}
    caminho = Path(nome_arquivo)
    if not caminho.is_file():
        logging.warning("Arquivo '%s' não encontrado.", nome_arquivo)
        return dados
    try:
        with open(caminho, 'r', encoding='utf-8', errors='replace') as f:
            for linha in f:
                linha = linha.rstrip('\n')
                if not linha or not linha[0].isdigit():
                    continue
                try:
                    ano = int(linha[0:4])
                    mes = int(linha[5:7])
                except ValueError:
                    continue
                
                def safe_float(start, end):
                    try:
                        s = linha[start:end].strip()
                        return float(s) if s != '' else None
                    except ValueError:
                        return None
               
                def safe_int(start, end):
                    try:
                        s = linha[start:end].strip()
                        return int(s) if s != '' else None
                    except ValueError:
                        return None

                fracao = safe_float(8, 16)
                num = safe_float(18, 23)
                desvio = safe_float(24, 29)
                nobs = safe_int(31, 35)
                dados.setdefault(ano, {})[mes] = (fracao, num, desvio, nobs)
    except OSError as e:
        logging.error("Erro ao ler '%s': %s", nome_arquivo, e)
    return dados

def ler_dados_diarios(nome_arquivo: str = "SN_d_tot_V2.0.txt") -> Dict[int, Dict[int, Dict[int, Tuple[float, Optional[int], Optional[float]]]]]:
    """
    Lê o arquivo de dados diários de manchas solares.
    Retorna: {ano: {mes: {dia: (fracao, num_manchas, desvio)}}}.
    Valores de '-1' no arquivo são convertidos para None.
    """
    dados = {}
    caminho = Path(nome_arquivo)
    if not caminho.is_file():
        logging.warning("Arquivo '%s' não encontrado.", nome_arquivo)
        return dados
    try:
        with open(caminho, 'r', encoding='utf-8', errors='replace') as f:
            for linha in f:
                linha = linha.rstrip('\n')
                if not linha or not linha[0].isdigit():
                    continue
                try:
                    ano = int(linha[0:4])
                    mes = int(linha[5:7])
                    dia = int(linha[8:10])
                except ValueError:
                    continue

                def safe_float(start, end):
                    try:
                        s = linha[start:end].strip()
                        return float(s) if s != '' else None
                    except ValueError:
                        return None
                
                def safe_int(start, end):
                    try:
                        s = linha[start:end].strip()
                        return int(s) if s != '' else None
                    except ValueError:
                        return None

                fracao = safe_float(11, 19)
                num = safe_int(21, 24)
                desvio = safe_float(25, 30)
                num_val = None if num == -1 else num
                dados.setdefault(ano, {}).setdefault(mes, {})[dia] = (fracao, num_val, desvio)
    except OSError as e:
        logging.error("Erro ao ler '%s': %s", nome_arquivo, e)
    return dados

def iterar_periodo_mensal(dados_mensais: Dict[int, Dict[int, Any]],
                          ano_inicio: int, mes_inicio: int,
                          ano_final: int, mes_final: int):
    """
    Gerador que itera sobre um intervalo de meses (inclusive).
    Yields: (ano, mes, dados_do_mes).
    Valida se os meses estão no intervalo [1, 12].
    """
    if mes_inicio < 1 or mes_inicio > 12 or mes_final < 1 or mes_final > 12:
        return
    start = ano_mes_to_total_months(ano_inicio, mes_inicio)
    end = ano_mes_to_total_months(ano_final, mes_final)
    if start > end:
        return
    for total in range(start, end + 1):
        a, m = total_months_to_ano_mes(total)
        if a in dados_mensais and m in dados_mensais[a]:
            yield a, m, dados_mensais[a][m]

def calcular_media_suavizada_robusta(dados_mensais: Dict[int, Dict[int, Tuple[float, Optional[float], Optional[float], Optional[int]]]],
                                      ano_inicio: int, ano_final: int) -> Dict[int, Dict[int, Optional[float]]]:
    """
    Calcula a média suavizada (fórmula SIDC) para o período solicitado.
    Utiliza pesos: 0.5 para os extremos da janela de 13 meses e 1.0 para os centrais.
    Retorna None para meses onde não há dados suficientes para o cálculo.
    """
    def obter(ano: int, mes: int) -> Optional[float]:
        val = dados_mensais.get(ano, {}).get(mes)
        if val is None:
            return None
        return val[1] 

    def index_offset(ano_base: int, mes_base: int, offset: int) -> Tuple[int, int]:
        total = ano_mes_to_total_months(ano_base, mes_base) + offset
        return total_months_to_ano_mes(total)

    suav = {}
    
    for ano in range(ano_inicio, ano_final + 1):
        suav[ano] = {}
        for mes in range(1, 13):
            soma = 0.0
            soma_pesos = 0.0
            
            for off, peso in [(-6, 0.5), (6, 0.5)]:
                a, m = index_offset(ano, mes, off)
                v = obter(a, m)
                if v is not None:
                    soma += v * peso
                    soma_pesos += peso
            
            for off in range(-5, 6):
                a, m = index_offset(ano, mes, off)
                v = obter(a, m)
                if v is not None:
                    soma += v
                    soma_pesos += 1.0
            
            suav[ano][mes] = (soma / soma_pesos) if soma_pesos > 0 else None
    return suav

def converter_dic_para_lista_suavizada(dicionario_suavizado: Dict[int, Dict[int, Optional[float]]],
                                       ano_inicio: int, mes_inicio: int, ano_final: int, mes_final: int) -> List[Tuple[float, float]]:
    """
    Converte o dicionário de médias suavizadas em uma lista ordenada de tuplas (data_decimal, valor).
    Remove entradas com valor None.
    """
    lista = []
    for a, m, _ in iterar_periodo_mensal(dicionario_suavizado, ano_inicio, mes_inicio, ano_final, mes_final):
        val = dicionario_suavizado.get(a, {}).get(m)
        if val is None:
            continue
        lista.append((ano_mes_para_decimal(a, m), val))
    return sorted(lista, key=lambda x: x[0])

def extrair_dados_por_data_decimal(dados_mensais: Dict[int, Dict[int, Tuple[float, Optional[float], Optional[float], Optional[int]]]],
                                   data_inicio_decimal: float, data_fim_decimal: float) -> List[float]:
    """
    Extrai uma lista de valores de manchas solares dentro de um intervalo decimal.
    Utilizada para isolar dados de um ciclo específico.
    """
    ano_i, mes_i = decimal_para_ano_mes(data_inicio_decimal)
    ano_f, mes_f = decimal_para_ano_mes(data_fim_decimal)
    lista = []
    for a, m, v in iterar_periodo_mensal(dados_mensais, ano_i, mes_i, ano_f, mes_f):
        data_dec = ano_mes_para_decimal(a, m)
        if data_dec < data_inicio_decimal - EPS or data_dec > data_fim_decimal + EPS:
            continue
        val = v[1]
        if val is not None:
            lista.append(val)
    return lista

def detectar_pontos_ciclo(lista_suavizada: List[Tuple[float, float]], metodo: str, janela_meses: int) -> List[Tuple[float, float]]:
    """
    Identifica pontos críticos (mínimos, máximos ou cruzamentos) na série suavizada.
    Parâmetros:
      metodo: String indicando a estratégia ('Mínimo', 'Máximo', 'Cruzamento').
      janela_meses: Inteiro definindo a vizinhança para detecção de extremos locais.
    
    Retorna: Lista de tuplas (data_decimal, valor).
    """
    pontos = []
    n = len(lista_suavizada)
    if n == 0:
        return []
    
    if 'Cruzamento' in metodo:
        media_periodo = sum(x[1] for x in lista_suavizada) / n
        for i in range(1, n):
            ant, atual = lista_suavizada[i-1][1], lista_suavizada[i][1]
            if 'Ascendente' in metodo and ant < media_periodo - EPS and atual >= media_periodo - EPS:
                pontos.append(lista_suavizada[i])
            elif 'Descendente' in metodo and ant > media_periodo + EPS and atual <= media_periodo + EPS:
                pontos.append(lista_suavizada[i])
        return pontos
    
    tipo = 'max' if 'Máximo' in metodo or 'Max' in metodo or 'MÁXIMO' in metodo else 'min'
    if n <= 2 * janela_meses:
        return []
    
    for i in range(janela_meses, n - janela_meses):
        central = lista_suavizada[i][1]
        viz = [x[1] for x in lista_suavizada[i - janela_meses:i + janela_meses + 1]]
        
        if tipo == 'max':
            if central + EPS >= max(viz):
                if not pontos or abs(pontos[-1][1] - central) > EPS:
                    pontos.append(lista_suavizada[i])
        else:
            if central - EPS <= min(viz):
                if not pontos or abs(pontos[-1][1] - central) > EPS:
                     pontos.append(lista_suavizada[i])
    return pontos

def executar_testes_estatisticos(dados_mensais: Dict[int, Dict[int, Tuple[float, Optional[float], Optional[float], Optional[int]]]],
                                 tipo_teste: str, **kwargs) -> str:
    """
    Realiza testes estatísticos (t-Student ou Qui-Quadrado) comparando ciclos.
    Requer as bibliotecas numpy e scipy. Retorna string com o relatório.
    """
    try:
        import numpy as np
        from scipy import stats
    except ImportError:
        return "Erro: As bibliotecas 'numpy' e 'scipy' são necessárias para esta análise."
    
    def _extrair_ciclo_oficial(n_ciclo: int) -> List[float]:
        info = next((c for c in LISTA_CICLOS_OFICIAIS if c[0] == n_ciclo), None)
        if not info:
            return []
        a0, m0, af, mf = info[1], info[2], info[3], info[4]
        if af is None:
            af, mf = max(dados_mensais.keys()) if dados_mensais else 0, 12
        vals = []
        for a, m, v in iterar_periodo_mensal(dados_mensais, a0, m0, af, mf):
            val = v[1]
            if val is not None:
                vals.append(val)
        return vals

    try:
        if tipo_teste == 't_student':
            a = kwargs.get('ciclo_a')
            b = kwargs.get('ciclo_b')
            da = _extrair_ciclo_oficial(a)
            db = _extrair_ciclo_oficial(b)
            if not da or not db:
                return "Dados insuficientes para um dos ciclos."
            estat, p = stats.ttest_ind(da, db, equal_var=False, nan_policy='omit')
            conclusao = "Diferença estatisticamente significativa." if p < 0.05 else "Sem diferença significativa."
            return f"Estatística t: {estat:.4f}\nValor-p: {p:.4f}\nConclusão: {conclusao}"
        
        elif tipo_teste in ('qui_quadrado', 'qui_quadrado_calculado'):
            if tipo_teste == 'qui_quadrado':
                ciclo = kwargs.get('ciclo')
                dados_ciclo = _extrair_ciclo_oficial(ciclo)
                todos = [v[1] for a in dados_mensais for m, v in dados_mensais[a].items() if v[1] is not None]
            else:
                dados_ciclo = kwargs.get('dados_ciclo', [])
                todos = kwargs.get('todos_dados_historicos', [])
            
            if not dados_ciclo or not todos:
                return "(Dados insuficientes)"
            
            bins = [0, 50, 150, float('inf')]
            hist_esperado, _ = np.histogram(todos, bins)
            prob_esperada = hist_esperado / len(todos)
            freq_observada, _ = np.histogram(dados_ciclo, bins)
            freq_esperada = prob_esperada * len(dados_ciclo)
            
            if any(freq_esperada < 5):
                try:
                    estat_ks, p_ks = stats.ks_2samp(dados_ciclo, todos)
                    conclusao = "Distribuição diferente (KS, p<0.05)" if p_ks < 0.05 else "Distribuição consistente (KS)"
                    return f"Teste KS (fallback): estat={estat_ks:.4f}, p={p_ks:.4f}.\nConclusão: {conclusao}"
                except Exception as e:
                    return f"Não é possível aplicar Qui-Quadrado nem KS: {e}"
            
            freq_esperada = freq_esperada + 1e-9
            freq_observada = freq_observada + 1e-9
        
            estat_chi2, p_chi2 = stats.chisquare(freq_observada, freq_esperada)
            conclusao = "Perfil diferente da norma histórica." if p_chi2 < 0.05 else "Perfil consistente com a norma."
            return f"Qui-Quadrado: {estat_chi2:.4f}\nValor-p: {p_chi2:.4f}\nConclusão: {conclusao}"
        else:
            return "Tipo de teste desconhecido."
    except Exception as e:
        return f"Erro ao executar teste: {e}"

def _obter_lista_dias_mes(dados_diarios, ano, mes) -> List[int]:
    """
    Auxiliar: retorna lista de contagens diárias válidas para um mês específico.
    """
    lista = []
    if ano in dados_diarios and mes in dados_diarios[ano]:
        for dia in sorted(dados_diarios[ano][mes].keys()):
            cont = dados_diarios[ano][mes][dia][1]
            if cont is None:
                continue
            lista.append(cont)
    return lista

def contar_dias_zero_mes(dados_diarios, ano, mes) -> int:
    """
    Retorna a contagem de dias com zero manchas solares em um determinado mês.
    """
    return sum(1 for v in _obter_lista_dias_mes(dados_diarios, ano, mes) if v == 0)

def calcular_stats_pelo_diario(dados_diarios, ano, mes) -> Tuple[float, float]:
    """
    Calcula média e desvio padrão a partir dos dados diários.
    Retorna (media, desvio_padrao).
    """
    lista = _obter_lista_dias_mes(dados_diarios, ano, mes)
    n = len(lista)
    if n == 0:
        return 0.0, 0.0
    media = sum(lista) / n
    if n > 1:
        var = sum((x - media) ** 2 for x in lista) / (n - 1)
        return media, math.sqrt(var)
    else:
        return media, 0.0

def plotar_grafico_base(dados_mensais, dados_suavizados, ano_inicio, mes_inicio, 
                        ano_final, mes_final, mostrar_pontos=True):
    """
    Gera a figura base do Matplotlib com os dados mensais (pontos azuis)
    e a curva suavizada (linha preta).
    """
    x_mensal, y_mensal = [], []
    for a, m, v in iterar_periodo_mensal(dados_mensais, ano_inicio, mes_inicio, ano_final, mes_final):
        val = v[1]
        if val is None:
            continue
        x_mensal.append(ano_mes_para_decimal(a, m))
        y_mensal.append(val)
    
    suav_map = {ano_mes_para_decimal(a, m): val for a, meses in dados_suavizados.items() for m, val in meses.items() if val is not None}
    y_suavizado = [suav_map.get(xd, None) for xd in x_mensal]
    
    fig, ax = plt.subplots(figsize=(10, 5), dpi=100)
    
    # MODIFICADO: Só plota os pontos se mostrar_pontos for True
    if mostrar_pontos:
        ax.plot(x_mensal, y_mensal, 'b.', markersize=2, alpha=0.4, label='Média Mensal (Original)')
    
    xs_plot, ys_plot = [], []
    for xd, yd in zip(x_mensal, y_suavizado):
        if yd is None:
            if xs_plot:
                ax.plot(xs_plot, ys_plot, 'k-', linewidth=1.5, alpha=0.9, label='Média Suavizada' if 'Média Suavizada' not in [l.get_label() for l in ax.lines] else None)
                xs_plot, ys_plot = [], []
            continue
        xs_plot.append(xd)
        ys_plot.append(yd)
    if xs_plot:
        ax.plot(xs_plot, ys_plot, 'k-', linewidth=1.5, alpha=0.9, label='Média Suavizada' if 'Média Suavizada' not in [l.get_label() for l in ax.lines] else None)
    
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('Ano')
    ax.set_ylabel('Número de Manchas')
    return fig, ax

# --- Inicialização dos Dados ---
try:
    dados_diarios = ler_dados_diarios()
    dados_mensais = ler_dados_mensais()
    if dados_diarios and dados_mensais:
        logging.info("Dados carregados com sucesso.")
    else:
        logging.warning("Dados vazios ou arquivos não encontrados; algumas funcionalidades ficarão limitadas.")
except Exception as e:
    logging.error("Erro ao carregar dados: %s", e)
    dados_diarios = {}
    dados_mensais = {}

# --- Interface Gráfica (Widgets) ---

# Aba 1: Análise
ano_inicio_analise = widgets.IntText(value=1818, description='Ano Ini', layout={'width': '180px'})
mes_inicio_analise = widgets.IntText(value=1, description='Mês Ini', layout={'width': '150px'})
ano_fim_analise = widgets.IntText(value=2025, description='Ano Fim', layout={'width': '180px'})
mes_fim_analise = widgets.IntText(value=12, description='Mês Fim', layout={'width': '150px'})

# NOVO WIDGET
chk_pontos_analise = widgets.Checkbox(value=True, description='Mostrar Pontos', indent=False)

btn_extremos = widgets.Button(description="Extremos")
btn_media_zero = widgets.Button(description="Meses com Média Zero")
output_analise_texto = widgets.Output(layout={'border': '1px solid black', 'height': '120px', 'overflow': 'scroll'})
output_analise_img = widgets.Output(layout={'border': '1px solid black', 'min_height': '400px'})

# Aba 2: Ciclos
dropdown_metodo_ciclo = widgets.Dropdown(
    options=['Mínimo a Mínimo', 'Máximo a Máximo', 'Cruzamento Ascendente', 'Cruzamento Descendente'],
    value='Mínimo a Mínimo',
    description='Método'
)
janela_ciclo_meses = widgets.IntText(value=36, description='Janela (meses)', layout={'width': '200px'})

# NOVO WIDGET
chk_pontos_ciclos = widgets.Checkbox(value=True, description='Mostrar Pontos', indent=False)

btn_calcular_ciclos = widgets.Button(description="Calcular e Comparar Ciclos")
output_ciclo_texto = widgets.Output(layout={'border': '1px solid black', 'height': '200px', 'overflow': 'scroll'})
output_ciclo_img = widgets.Output(layout={'border': '1px solid black', 'min_height': '400px'})

# Aba 3: Estatística
input_ciclo_a = widgets.IntText(value=23, description='Ciclo A', layout={'width': '150px'})
input_ciclo_b = widgets.IntText(value=24, description='Ciclo B', layout={'width': '150px'})
btn_teste_t = widgets.Button(description="Teste t-Student (Oficiais)")

input_ciclo_chi = widgets.IntText(value=24, description='Ciclo', layout={'width': '150px'})
btn_teste_chi = widgets.Button(description="Teste Qui-Quad (Oficial)")
output_estatistica = widgets.Output(layout={'border': '1px solid black', 'height': '150px', 'overflow': 'scroll'})

# Aba 4: Consultas
input_ano_consulta = widgets.IntText(value=2024, description='Ano:', layout={'width': '180px'})
btn_consulta_anual = widgets.Button(description="Média Anual")

input_ano_mes = widgets.IntText(value=2024, description='Ano', layout={'width': '180px'})
input_mes_mes = widgets.IntText(value=1, description='Mês', layout={'width': '140px'})
btn_consulta_mensal = widgets.Button(description="Média Mensal")

input_ano_dia = widgets.IntText(value=2024, description='Ano', layout={'width': '180px'})
input_mes_dia = widgets.IntText(value=1, description='Mês', layout={'width': '140px'})
input_dia_dia = widgets.IntText(value=1, description='Dia', layout={'width': '140px'})
btn_consulta_dia = widgets.Button(description="Valor Diário")

input_ano_stat = widgets.IntText(value=2024, description='Ano', layout={'width': '180px'})
input_mes_stat = widgets.IntText(value=1, description='Mês', layout={'width': '140px'})
btn_consulta_stat = widgets.Button(description="Calcular Média/DP (Diário)")

input_ano_zero = widgets.IntText(value=2024, description='Ano', layout={'width': '180px'})
input_mes_zero = widgets.IntText(value=1, description='Mês', layout={'width': '140px'})
btn_consulta_zero = widgets.Button(description="Contar Dias Sem Manchas")

output_consulta = widgets.Output(layout={'border': '1px solid black', 'height': '200px', 'overflow': 'scroll'})

def on_click_extremos(botao):
    output_analise_texto.clear_output()
    output_analise_img.clear_output()
    ano_i, mes_i, ano_f, mes_f = ano_inicio_analise.value, mes_inicio_analise.value, ano_fim_analise.value, mes_fim_analise.value
    
    if mes_i < 1 or mes_i > 12 or mes_f < 1 or mes_f > 12:
        with output_analise_texto:
            print("Entrada inválida: mês deve estar entre 1 e 12.")
        return
    
    max_diario = -1
    data_max_diario = (None, None, None)
    for a in range(ano_i, ano_f + 1):
        if a not in dados_diarios:
            continue
        mi = mes_i if a == ano_i else 1
        mf = mes_f if a == ano_f else 12
        for m in range(mi, mf + 1):
            if m not in dados_diarios[a]:
                continue
            for d, tup in dados_diarios[a][m].items():
                val = tup[1]
                if val is None:
                    continue
                if val > max_diario:
                    max_diario = val
                    data_max_diario = (a, m, d)
    
    valor_max_mensal = -1
    data_max_mensal = (None, None)
    valor_min_mensal = float('inf')
    data_min_mensal = (None, None)
    for a, m, v in iterar_periodo_mensal(dados_mensais, ano_i, mes_i, ano_f, mes_f):
        num = v[1]
        if num is None:
            continue
        if num > valor_max_mensal:
            valor_max_mensal = num
            data_max_mensal = (a, m)
        if num < valor_min_mensal:
            valor_min_mensal = num
            data_min_mensal = (a, m)
            
    resultados = {
        "max_diario": (data_max_diario, max_diario if max_diario >= 0 else None),
        "max_mensal": (data_max_mensal, valor_max_mensal if valor_max_mensal != -1 else None),
        "min_mensal": (data_min_mensal, valor_min_mensal if valor_min_mensal != float('inf') else None)
    }
    with output_analise_texto:
        print(f"Máximo Diário: {resultados['max_diario'][1]} (em {resultados['max_diario'][0]})")
        print(f"Máximo Mensal: {resultados['max_mensal'][1]} (em {resultados['max_mensal'][0]})")
        print(f"Mínimo Mensal: {resultados['min_mensal'][1]} (em {resultados['min_mensal'][0]})")
    with output_analise_img:
        dados_suavizados_periodo = calcular_media_suavizada_robusta(dados_mensais, ano_i - 1, ano_f + 1)
        # MODIFICADO: Passa o valor do checkbox
        fig, ax = plotar_grafico_base(dados_mensais, dados_suavizados_periodo, ano_i, mes_i, ano_f, mes_f, mostrar_pontos=chk_pontos_analise.value)
        maximo = resultados['max_mensal']
        minimo = resultados['min_mensal']
        if maximo[0][0] is not None:
            ax.plot(ano_mes_para_decimal(maximo[0][0], maximo[0][1]), maximo[1], 'r*', ms=12, label='Máx Mensal')
        if minimo[0][0] is not None:
            ax.plot(ano_mes_para_decimal(minimo[0][0], minimo[0][1]), minimo[1], 'bv', ms=10, label='Mín Mensal')
        ax.legend()
        plt.show()

def encontrar_meses_media_zero(dados_mensais, ano_inicio, mes_inicio, ano_final, mes_final):
    """
    Localiza meses onde a média de manchas solares registrada foi exatamente zero.
    """
    meses = []
    for a, m, v in iterar_periodo_mensal(dados_mensais, ano_inicio, mes_inicio, ano_final, mes_final):
        val = v[1]
        if val is not None and val == 0.0:
            meses.append(((a, m), val))
    return meses

def on_click_media_zero(botao):
    output_analise_texto.clear_output()
    output_analise_img.clear_output()
    ano_i, mes_i, ano_f, mes_f = ano_inicio_analise.value, mes_inicio_analise.value, ano_fim_analise.value, mes_fim_analise.value
    todos = encontrar_meses_media_zero(dados_mensais, ano_i, mes_i, ano_f, mes_f)
    with output_analise_texto:
        print(f"Total de meses com média 0.0 encontrados: {len(todos)}")
        if todos:
            print("Exibindo os 10 primeiros (se houver):")
            for (data, contagem) in todos[:10]:
                print(f"  - {data[1]}/{data[0]}")
    with output_analise_img:
        dados_suavizados_periodo = calcular_media_suavizada_robusta(dados_mensais, ano_i - 1, ano_f + 1)
        # MODIFICADO: Passa o valor do checkbox
        fig, ax = plotar_grafico_base(dados_mensais, dados_suavizados_periodo, ano_i, mes_i, ano_f, mes_f, mostrar_pontos=chk_pontos_analise.value)
        if todos:
            added = False
            for (data, contagem) in todos:
                x = ano_mes_para_decimal(data[0], data[1])
                y = contagem
                label = "Mês com Média 0.0" if not added else None
                added = True
                ax.plot(x, y, 'o', color='black', ms=8, mfc='none', mew=2, label=label)
            ax.legend()
        plt.show()

def on_click_calcular_ciclos(botao):
    output_ciclo_texto.clear_output()
    output_ciclo_img.clear_output()
    ano_i, mes_i = ano_inicio_analise.value, mes_inicio_analise.value
    ano_f, mes_f = ano_fim_analise.value, mes_fim_analise.value
    metodo = dropdown_metodo_ciclo.value
    janela = janela_ciclo_meses.value
    margem_anual = (janela // 12) + 2
    dados_suavizados_completos = calcular_media_suavizada_robusta(dados_mensais, ano_i - margem_anual, ano_f + margem_anual)
    lista = converter_dic_para_lista_suavizada(dados_suavizados_completos, ano_i, mes_i, ano_f, mes_f)
    if not lista:
        with output_ciclo_texto:
            print("Dados insuficientes no período selecionado.")
        return
    pontos = detectar_pontos_ciclo(lista, metodo, janela)
    todos_historico = [v[1] for a in dados_mensais for m, v in dados_mensais[a].items() if v[1] is not None]
    with output_ciclo_texto:
        print(f"--- Ciclos Calculados ({metodo}) ---")
        if len(pontos) < 2:
            print("Não foram detectados ciclos suficientes para análise.")
        for i in range(len(pontos) - 1):
            di = pontos[i][0]
            df = pontos[i + 1][0]
            dados_deste_ciclo = extrair_dados_por_data_decimal(dados_mensais, di, df)
            if dados_deste_ciclo and todos_historico:
                resultado_teste = executar_testes_estatisticos(
                    dados_mensais,
                    tipo_teste='qui_quadrado_calculado',
                    dados_ciclo=dados_deste_ciclo,
                    todos_dados_historicos=todos_historico,
                )
            else:
                resultado_teste = "(Sem dados para teste)"
            print(f"  {di:.2f} a {df:.2f} (Duração: {df - di:.2f} anos) {resultado_teste}")
        print("\n--- Ciclos Oficiais (Min-Min) que interceptam o período ---")
        for ciclo_oficial in LISTA_CICLOS_OFICIAIS:
            ciclo_ano_inicio = ciclo_oficial[1]
            ciclo_ano_fim = ciclo_oficial[3] if ciclo_oficial[3] else ano_f
            if ciclo_ano_inicio <= ano_f and ciclo_ano_fim >= ano_i:
                texto_fim = f"{ciclo_oficial[3] + ciclo_oficial[4]/12:.2f}" if ciclo_oficial[3] else "Ativo"
                print(f"  Ciclo {ciclo_oficial[0]}: {ciclo_oficial[1] + ciclo_oficial[2]/12:.2f} a {texto_fim}")
    with output_ciclo_img:
        # MODIFICADO: Passa o valor do checkbox
        fig, ax = plotar_grafico_base(dados_mensais, dados_suavizados_completos, ano_i, mes_i, ano_f, mes_f, mostrar_pontos=chk_pontos_ciclos.value)
        cor = 'k' if 'Mínimo' in metodo else ('r' if 'Máximo' in metodo else 'g')
        label_calculado = True
        for ponto in pontos:
            ax.axvline(ponto[0], color=cor, linestyle='--', alpha=0.7, label="Calculado" if label_calculado else None)
            ax.plot(ponto[0], ponto[1], cor + 'o')
            label_calculado = False
        label_oficial = True
        for ciclo_oficial in LISTA_CICLOS_OFICIAIS:
            ciclo_inicio_decimal = ciclo_oficial[1] + (ciclo_oficial[2] - 1) / 12.0
            if ciclo_inicio_decimal >= ano_i and ciclo_inicio_decimal <= ano_f + 1:
                ax.axvline(ciclo_inicio_decimal, color='b', linestyle=':', alpha=0.5, label="Oficial (Início)" if label_oficial else None)
                label_oficial = False
        ax.legend()
        plt.show()

def on_click_teste_t(botao):
    output_estatistica.clear_output()
    with output_estatistica:
        res = executar_testes_estatisticos(dados_mensais, 't_student', ciclo_a=input_ciclo_a.value, ciclo_b=input_ciclo_b.value)
        print(res)

def on_click_teste_chi(botao):
    output_estatistica.clear_output()
    with output_estatistica:
        res = executar_testes_estatisticos(dados_mensais, 'qui_quadrado', ciclo=input_ciclo_chi.value)
        print(res)

def on_click_consulta_anual(botao):
    output_consulta.clear_output()
    ano = input_ano_consulta.value
    with output_consulta:
        if ano in dados_mensais:
            valores = [v[1] for v in dados_mensais[ano].values() if v[1] is not None]
            if valores:
                media = sum(valores) / len(valores)
                print(f"Média de manchas em {ano}: {media:.2f}")
            else:
                print(f"Sem valores válidos para o ano {ano}.")
        else:
            print(f"Sem dados mensais para o ano {ano}.")

def on_click_consulta_mensal(botao):
    output_consulta.clear_output()
    ano, mes = input_ano_mes.value, input_mes_mes.value
    with output_consulta:
        if ano in dados_mensais and mes in dados_mensais[ano] and dados_mensais[ano][mes][1] is not None:
            print(f"Média de manchas em {mes}/{ano}: {dados_mensais[ano][mes][1]:.2f}")
        else:
            print(f"Sem dados válidos para {mes}/{ano}.")

def on_click_consulta_dia(botao):
    output_consulta.clear_output()
    ano, mes, dia = input_ano_dia.value, input_mes_dia.value, input_dia_dia.value
    with output_consulta:
        dd = dados_diarios.get(ano, {}).get(mes, {}).get(dia)
        if dd:
            obs = dd[1]
            if obs is None:
                print(f"Data {ano}-{mes}-{dia}: Sem observação (-1).")
            else:
                print(f"Manchas em {ano}-{mes}-{dia}: {obs}")
        else:
            print(f"Data {ano}-{mes}-{dia} não encontrada nos registros.")

def on_click_consulta_stat(botao):
    output_consulta.clear_output()
    ano, mes = input_ano_stat.value, input_mes_stat.value
    with output_consulta:
        media, dp = calcular_stats_pelo_diario(dados_diarios, ano, mes)
        lista = _obter_lista_dias_mes(dados_diarios, ano, mes)
        if not lista:
            print(f"Sem dados diários válidos para {mes}/{ano}.")
        else:
            print(f"Análise de {mes}/{ano} (calculado do diário):")
            print(f"  Média: {media:.2f}")
            print(f"  Desvio Padrão (amostral): {dp:.2f}")

def on_click_consulta_zero(botao):
    output_consulta.clear_output()
    ano, mes = input_ano_zero.value, input_mes_zero.value
    with output_consulta:
        cont = contar_dias_zero_mes(dados_diarios, ano, mes)
        print(f"Resultado para {mes}/{ano}:")
        print(f"  Total de dias sem manchas (contagem=0): {cont} dias.")

# Associação de Eventos
btn_extremos.on_click(on_click_extremos)
btn_media_zero.on_click(on_click_media_zero)
btn_calcular_ciclos.on_click(on_click_calcular_ciclos)
btn_teste_t.on_click(on_click_teste_t)
btn_teste_chi.on_click(on_click_teste_chi)
btn_consulta_anual.on_click(on_click_consulta_anual)
btn_consulta_mensal.on_click(on_click_consulta_mensal)
btn_consulta_dia.on_click(on_click_consulta_dia)
btn_consulta_stat.on_click(on_click_consulta_stat)
btn_consulta_zero.on_click(on_click_consulta_zero)

# Layout Final (Abas)
# MODIFICADO: Incluindo os checkboxes no layout
aba_analise = widgets.VBox([
    widgets.HBox([ano_inicio_analise, mes_inicio_analise, ano_fim_analise, mes_fim_analise]),
    widgets.HBox([btn_extremos, btn_media_zero, chk_pontos_analise]),
    output_analise_texto,
    output_analise_img
])

aba_ciclos = widgets.VBox([
    widgets.HBox([dropdown_metodo_ciclo, janela_ciclo_meses, chk_pontos_ciclos]),
    btn_calcular_ciclos,
    output_ciclo_texto,
    output_ciclo_img
])

aba_estatistica = widgets.VBox([
    widgets.Label("Comparar médias de dois Ciclos Oficiais (Teste T):"),
    widgets.HBox([input_ciclo_a, input_ciclo_b, btn_teste_t]),
    widgets.Label("Verificar consistência de um Ciclo Oficial (Qui-Quadrado):"),
    widgets.HBox([input_ciclo_chi, btn_teste_chi]),
    output_estatistica
])

aba_consultas = widgets.VBox([
    widgets.Label("Média Anual (do arquivo mensal):"),
    widgets.HBox([input_ano_consulta, btn_consulta_anual]),
    widgets.Label("Média Mensal (do arquivo mensal):"),
    widgets.HBox([input_ano_mes, input_mes_mes, btn_consulta_mensal]),
    widgets.Label("Valor por Dia Específico (do arquivo diário):"),
    widgets.HBox([input_ano_dia, input_mes_dia, input_dia_dia, btn_consulta_dia]),
    widgets.Label("Contar Dias Sem Manchas (calculado do diário):"),
    widgets.HBox([input_ano_zero, input_mes_zero, btn_consulta_zero]),
    widgets.Label("Média/DP do Mês (calculado do diário):"),
    widgets.HBox([input_ano_stat, input_mes_stat, btn_consulta_stat]),
    output_consulta
])

app = widgets.Tab([aba_analise, aba_ciclos, aba_estatistica, aba_consultas])
app.set_title(0, 'Análise de Período')
app.set_title(1, 'Ciclos Solares')
app.set_title(2, 'Estatística (Ciclos Oficiais)')
app.set_title(3, 'Consultas')

display(app)
```
